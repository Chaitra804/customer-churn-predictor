```python
import pandas as pd

train_df = pd.read_csv("../data/raw/train.csv")
test_df = pd.read_csv("../data/raw/test.csv")

print("Train:", train_df.shape)
print("Test:", test_df.shape)
```

```python
train = train_df.copy()
test = test_df.copy()
```

```python
print(train.head())
print(test.head())
```

```python
drop_cols = ['Customer_ID', 'Name', 'Email']

train = train.drop(drop_cols, axis=1)
test = test.drop(drop_cols, axis=1)
```

```python
mapping = {'Daily': 3, 'Weekly': 2, 'Monthly': 1}

train['Login_Frequency'] = train['Login_Frequency'].map(mapping)
test['Login_Frequency'] = test['Login_Frequency'].map(mapping)
```

```python
for df in [train, test]:
    df['ticket_length'] = df['Last_Support_Ticket'].apply(lambda x: len(str(x)))
    df['engagement_score'] = df['Daily_Usage_Mins'] * df['Login_Frequency']
    df['has_complaint'] = df['Last_Support_Ticket'].str.contains(
        'issue|error|problem|not working', case=False, na=False
    ).astype(int)
```

```python
train = train.drop(['Last_Support_Ticket'], axis=1)
test = test.drop(['Last_Support_Ticket'], axis=1)
```

```python
train = train.fillna(0)
test = test.fillna(0)
```

```python
train.to_csv("../data/processed/train_processed.csv", index=False)
test.to_csv("../data/processed/test_processed.csv", index=False)
```
